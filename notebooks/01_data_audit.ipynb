```python
data_dir = data_path.parent

coverage_records = []

for year in range(1981, 2027):
    yearly_matches = pd.read_csv(data_dir / f"atp_matches_{year}.csv")

    record = {
        "year": year,
        "matches": len(yearly_matches),
    }

    record.update(
        yearly_matches.notna().mean().mul(100).to_dict()
    )

    coverage_records.append(record)

coverage_by_year = pd.DataFrame(coverage_records)
```

```python
reference_columns = matches_1981.columns.tolist()
schema_issues = {}

for year in range(1981, 2027):
    columns = pd.read_csv(
        data_dir / f"atp_matches_{year}.csv",
        nrows=0,
    ).columns.tolist()

    if columns != reference_columns:
        schema_issues[year] = {
            "missing": sorted(set(reference_columns) - set(columns)),
            "extra": sorted(set(columns) - set(reference_columns)),
            "order_changed": set(columns) == set(reference_columns),
        }

schema_issues
```

```python
yearly_frames = []

for year in range(1981, 2027):
    yearly_matches = pd.read_csv(
        data_dir / f"atp_matches_{year}.csv"
    )

    yearly_matches["source_year"] = year
    yearly_frames.append(yearly_matches)

matches = pd.concat(
    yearly_frames,
    ignore_index=True,
)
```

```python
matches["tourney_date"] = pd.to_datetime(
    matches["tourney_date"].astype(str),
    format="%Y%m%d",
)
```

```python
core_level_mask = matches["tourney_level"].isin(
    ["A", "M", "G", "F"]
)

prediction_surface_mask = matches["surface"].isin(
    ["Hard", "Clay", "Grass"]
)

state_matches = matches.loc[
    core_level_mask
].copy()

model_eligible_matches = matches.loc[
    core_level_mask & prediction_surface_mask
].copy()
```

```python
matches.shape, state_matches.shape, model_eligible_matches.shape
```

```python
score_upper = (
    state_matches["score"]
    .fillna("")
    .str.upper()
)

match_status = pd.Series(
    "completed",
    index=state_matches.index,
    dtype="string",
)

match_status[score_upper.eq("")] = "missing_score"

match_status[
    score_upper.str.contains(r"W/O|WALKOVER", regex=True)
] = "walkover"

match_status[
    score_upper.str.contains(r"\bRET\b", regex=True)
] = "retirement"

match_status[
    score_upper.str.contains(r"\bDEF\b", regex=True)
] = "default"

match_status[
    score_upper.str.contains(
        r"ABD|ABN|UNFINISHED|IN PROGRESS",
        regex=True,
    )
] = "abandoned"

match_status.value_counts()
```

```python
state_matches["match_status"] = match_status
```

```python
model_eligible_matches = state_matches.loc[
    state_matches["surface"].isin(["Hard", "Clay", "Grass"])
    & state_matches["match_status"].eq("completed")
].copy()
```

```python
state_matches.shape, model_eligible_matches.shape
```

```python
full_row_duplicates = state_matches.duplicated().sum()

key_columns = ["tourney_id", "match_num"]

duplicate_key_rows = state_matches.loc[
    state_matches.duplicated(
        subset=key_columns,
        keep=False,
    )
].sort_values(key_columns)

full_row_duplicates, len(duplicate_key_rows)
```

```python
round_order_map = {
    "ER": 0,
    "RR": 0,
    "R128": 1,
    "R64": 2,
    "R32": 3,
    "R16": 4,
    "QF": 5,
    "SF": 6,
    "BR": 7,
    "F": 8,
}

state_matches["round_order"] = (
    state_matches["round"]
    .map(round_order_map)
)
```

```python
state_matches.loc[
    state_matches["round_order"].isna(),
    "round",
].value_counts(dropna=False)
```

```python
state_matches["source_row_order"] = state_matches.index

state_matches = (
    state_matches
    .sort_values(
        [
            "tourney_date",
            "tourney_id",
            "round_order",
            "match_num",
            "source_row_order",
        ],
        kind="stable",
    )
    .reset_index(drop=True)
)
```
